## Exercise 1: Expense analysis

You have 4 weeks × 7 days of spending. Rows are weeks, columns are weekdays.

```python
rng = np.random.default_rng(0)
spend = rng.integers(100,2000, size=(4,7))
print(spend)
```

### 1a

Compute the **total per week** and the **average per weekday** across the 4 weeks.

*Hint:* choose the right `axis` for each. `axis=0` is the dimension that disappears.

```python
total_per_week = spend.sum(axis=1,keepdims=True)
average_per_day = spend.mean(axis=0)
print(total_per_week)
print(average_per_day)
```

### 1b

Find every amount **over 1500**, and **how many** there are.

```python
over_1500 = spend[spend > 1500]
print(over_1500)
print(over_1500.size)
```

### 1c

Create a copy where every amount over 1500 is **capped at 1500**.

*Hint:* `np.clip` or `np.minimum`. Only apply the constraint the requirement asks for.

```python
copy = np.clip(spend,None,1500)
print(copy)
```

### 1d

Express each day's amount as a **percentage of its week's total**, giving a `(4, 7)` result. Then prove every row adds up to 100.

*Hints:* broadcasting compares shapes from the right; don't hardcode the number of weeks (`keepdims=True`); compare floats with `np.allclose`, not `==`.

```python
percent_spending = spend * 100 / total_per_week
print(percent_spending)

print(np.allclose(percent_spending.sum(axis=1),100))
```

## Exercise 2: Tiny semantic search

Five "documents" with 3-dimensional embeddings, and a query embedding.

```python
docs = np.array([[0.9, 0.1, 0.0],   # "refund policy"
                 [0.8, 0.2, 0.1],   # "return an item"
                 [0.0, 0.9, 0.3],   # "shipping times"
                 [0.1, 0.1, 0.9],   # "reset password"
                 [0.0, 0.8, 0.5]])  # "track my order"
query = np.array([0.85, 0.15, 0.05])  # "how do I get my money back"
```

### 2a

Score every document against the query with a single `@`.

```python
match_score = docs @ query
print(match_score)
```

### 2b

Return the indices of the **top 2** documents, best first.

*Hint:* `np.argsort`, then think about how to reverse the order.

```python
top_match_indices = np.argsort(match_score,descending=True)[:2]
print(top_match_indices)
```
